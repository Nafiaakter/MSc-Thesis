```python
!pip -q install openpyxl statsmodels scipy seaborn matplotlib pandas numpy

import os
import zipfile
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import (
    shapiro,
    ttest_ind,
    ttest_rel,
    mannwhitneyu,
    wilcoxon
)

import statsmodels.api as sm
from statsmodels.formula.api import ols

from IPython.display import display

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

print("Packages imported successfully.")


from google.colab import files

print("Please upload your Excel dataset.")
uploaded = files.upload()

uploaded_filename = list(uploaded.keys())[0]

print("\nUploaded file:")
print(uploaded_filename)


SHEET_NAME = "50_Participants"

df = pd.read_excel(
    uploaded_filename,
    sheet_name=SHEET_NAME
)

print("\nDataset loaded successfully.")
print("Shape:", df.shape)

print("\nColumns:")
for col in df.columns:
    print("-", col)



required_columns = [
    "Participant_Code",
    "Group",
    "Pre_Score_10",
    "Post_Score_10",
    "Retention_Score_10",
    "Pre_Time_sec",
    "Post_Time_sec",
    "Retention_Time_sec",
    "Pre_Engagement_1to5",
    "Post_Engagement_1to5",
    "Retention_Engagement_1to5"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        "The following required columns are missing:\n"
        + "\n".join(missing_columns)
    )

print("\nAll required columns are present.")



df["Group"] = (
    df["Group"]
    .astype(str)
    .str.strip()
    .str.casefold()
    .map({
        "ai": "AI",
        "control": "Control"
    })
)

if df["Group"].isna().any():

    bad_rows = df[df["Group"].isna()]

    print("\nUnexpected group labels were found:")
    print(bad_rows)

    raise ValueError(
        "\nUnexpected group labels found. "
        "Please check the Group column in the Excel file."
    )



group_counts = df["Group"].value_counts()

print("\n============================================================")
print("PARTICIPANT COUNTS")
print("============================================================")

print(group_counts)


if set(df["Group"].unique()) != {"AI", "Control"}:
    raise ValueError(
        "The dataset must contain both AI and Control groups."
    )

if (
    group_counts.get("AI", 0) != 25
    or
    group_counts.get("Control", 0) != 25
):
    print("\nWARNING:")
    print(
        "The code expected 25 AI and 25 Control participants."
    )
    print(
        "The analysis will continue, but please verify your dataset."
    )
else:
    print("\n✓ 25 AI participants")
    print("✓ 25 Control participants")



print("\n============================================================")
print("FIRST 10 PARTICIPANTS")
print("============================================================")

display(df.head(10))



duplicate_codes = df[
    df["Participant_Code"].duplicated(keep=False)
]

if len(duplicate_codes) > 0:

    print("\nWARNING: Duplicate participant codes found:")
    display(duplicate_codes)

else:

    print("\n✓ No duplicate participant codes found.")



print("\n============================================================")
print("MISSING DATA")
print("============================================================")

missing_summary = df[required_columns].isnull().sum()

display(
    missing_summary
    .to_frame("Missing_Count")
)

if missing_summary.sum() == 0:
    print("✓ No missing values in required variables.")
else:
    print("WARNING: Missing values detected.")



numeric_columns = [
    "Pre_Score_10",
    "Post_Score_10",
    "Retention_Score_10",
    "Pre_Time_sec",
    "Post_Time_sec",
    "Retention_Time_sec",
    "Pre_Engagement_1to5",
    "Post_Engagement_1to5",
    "Retention_Engagement_1to5"
]

for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )



print("\n============================================================")
print("RANGE CHECKS")
print("============================================================")

score_columns = [
    "Pre_Score_10",
    "Post_Score_10",
    "Retention_Score_10"
]

for col in score_columns:

    invalid = df[
        (df[col] < 0) |
        (df[col] > 10)
    ]

    if len(invalid) > 0:

        print(
            f"WARNING: {len(invalid)} invalid values in {col}"
        )

    else:

        print(f"✓ {col}: all values 0-10")


engagement_columns = [
    "Pre_Engagement_1to5",
    "Post_Engagement_1to5",
    "Retention_Engagement_1to5"
]

for col in engagement_columns:

    invalid = df[
        (df[col] < 1) |
        (df[col] > 5)
    ]

    if len(invalid) > 0:

        print(
            f"WARNING: {len(invalid)} invalid values in {col}"
        )

    else:

        print(f"✓ {col}: all values 1-5")


time_columns = [
    "Pre_Time_sec",
    "Post_Time_sec",
    "Retention_Time_sec"
]

for col in time_columns:

    invalid = df[df[col] <= 0]

    if len(invalid) > 0:

        print(
            f"WARNING: {len(invalid)} invalid values in {col}"
        )

    else:

        print(f"✓ {col}: all values positive")



df["Score_Improvement"] = (
    df["Post_Score_10"]
    -
    df["Pre_Score_10"]
)


df["Retention_Change"] = (
    df["Retention_Score_10"]
    -
    df["Post_Score_10"]
)


df["Time_Reduction_sec"] = (
    df["Pre_Time_sec"]
    -
    df["Post_Time_sec"]
)


df["Retention_Time_Change_sec"] = (
    df["Retention_Time_sec"]
    -
    df["Post_Time_sec"]
)


df["Engagement_Improvement"] = (
    df["Post_Engagement_1to5"]
    -
    df["Pre_Engagement_1to5"]
)


df["Retention_Engagement_Change"] = (
    df["Retention_Engagement_1to5"]
    -
    df["Post_Engagement_1to5"]
)

df["Retention_Percent_of_Post"] = np.where(
    df["Post_Score_10"] != 0,
    (
        df["Retention_Score_10"]
        /
        df["Post_Score_10"]
    ) * 100,
    np.nan
)



cleaned_data_file = "cleaned_thesis_dataset.csv"

df.to_csv(
    cleaned_data_file,
    index=False
)

print("\nCleaned dataset saved as:")
print(cleaned_data_file)



print("\n============================================================")
print("DESCRIPTIVE STATISTICS")
print("============================================================")


def descriptive_statistics(
    data,
    variable_columns
):

    rows = []

    for group in ["AI", "Control"]:

        subset = data[
            data["Group"] == group
        ]

        for variable in variable_columns:

            values = subset[variable].dropna()

            rows.append({
                "Group": group,
                "Variable": variable,
                "N": len(values),
                "Mean": values.mean(),
                "SD": values.std(ddof=1),
                "Median": values.median(),
                "Min": values.min(),
                "Max": values.max()
            })

    return pd.DataFrame(rows)


descriptive_variables = [
    "Pre_Score_10",
    "Post_Score_10",
    "Retention_Score_10",
    "Pre_Time_sec",
    "Post_Time_sec",
    "Retention_Time_sec",
    "Pre_Engagement_1to5",
    "Post_Engagement_1to5",
    "Retention_Engagement_1to5"
]

descriptive_results = descriptive_statistics(
    df,
    descriptive_variables
)

display(
    descriptive_results.round(3)
)

descriptive_results.to_csv(
    "descriptive_statistics.csv",
    index=False
)



summary_rows = []

for group in ["AI", "Control"]:

    subset = df[
        df["Group"] == group
    ]

    summary_rows.append({

        "Group": group,

        "Pre Score Mean":
            subset["Pre_Score_10"].mean(),

        "Pre Score SD":
            subset["Pre_Score_10"].std(),

        "Post Score Mean":
            subset["Post_Score_10"].mean(),

        "Post Score SD":
            subset["Post_Score_10"].std(),

        "Retention Score Mean":
            subset["Retention_Score_10"].mean(),

        "Retention Score SD":
            subset["Retention_Score_10"].std(),

        "Pre Time Mean":
            subset["Pre_Time_sec"].mean(),

        "Post Time Mean":
            subset["Post_Time_sec"].mean(),

        "Retention Time Mean":
            subset["Retention_Time_sec"].mean(),

        "Pre Engagement Mean":
            subset["Pre_Engagement_1to5"].mean(),

        "Post Engagement Mean":
            subset["Post_Engagement_1to5"].mean(),

        "Retention Engagement Mean":
            subset["Retention_Engagement_1to5"].mean()
    })


summary_table = pd.DataFrame(
    summary_rows
)

print("\nSummary table:")
display(summary_table.round(3))

summary_table.to_csv(
    "group_summary.csv",
    index=False
)



print("\n============================================================")
print("SHAPIRO-WILK NORMALITY TESTS")
print("============================================================")


normality_variables = [
    "Pre_Score_10",
    "Post_Score_10",
    "Retention_Score_10",
    "Score_Improvement",
    "Retention_Change",
    "Pre_Time_sec",
    "Post_Time_sec",
    "Retention_Time_sec",
    "Time_Reduction_sec",
    "Pre_Engagement_1to5",
    "Post_Engagement_1to5",
    "Retention_Engagement_1to5",
    "Engagement_Improvement",
    "Retention_Engagement_Change"
]

normality_rows = []

for group in ["AI", "Control"]:

    subset = df[
        df["Group"] == group
    ]

    for variable in normality_variables:

        values = subset[variable].dropna()

        if len(values) >= 3:

            statistic, p_value = shapiro(values)

            normality_rows.append({

                "Group": group,
                "Variable": variable,
                "N": len(values),
                "Shapiro_W": statistic,
                "p_value": p_value,
                "Normal_at_0.05":
                    "Yes" if p_value > 0.05 else "No"
            })


normality_results = pd.DataFrame(
    normality_rows
)

display(
    normality_results.round(4)
)

normality_results.to_csv(
    "normality_tests.csv",
    index=False
)



def cohens_d_independent(
    group1,
    group2
):

    group1 = np.asarray(group1.dropna())
    group2 = np.asarray(group2.dropna())

    n1 = len(group1)
    n2 = len(group2)

    mean1 = np.mean(group1)
    mean2 = np.mean(group2)

    sd1 = np.std(group1, ddof=1)
    sd2 = np.std(group2, ddof=1)

    pooled_sd = np.sqrt(
        (
            (n1 - 1) * sd1**2
            +
            (n2 - 1) * sd2**2
        )
        /
        (n1 + n2 - 2)
    )

    if pooled_sd == 0:
        return np.nan

    return (
        (mean1 - mean2)
        /
        pooled_sd
    )


def cohens_dz(
    before,
    after
):

    before = np.asarray(before)
    after = np.asarray(after)

    differences = after - before

    sd_difference = np.std(
        differences,
        ddof=1
    )

    if sd_difference == 0:
        return np.nan

    return (
        np.mean(differences)
        /
        sd_difference
    )


def rank_biserial_from_u(
    u_statistic,
    n1,
    n2
):

    return (
        (2 * u_statistic)
        /
        (n1 * n2)
    ) - 1


print("\n============================================================")
print("BETWEEN-GROUP COMPARISONS")
print("============================================================")


between_variables = [

    (
        "Post Score",
        "Post_Score_10"
    ),

    (
        "Retention Score",
        "Retention_Score_10"
    ),

    (
        "Score Improvement",
        "Score_Improvement"
    ),

    (
        "Post Completion Time",
        "Post_Time_sec"
    ),

    (
        "Time Reduction",
        "Time_Reduction_sec"
    ),

    (
        "Post Engagement",
        "Post_Engagement_1to5"
    ),

    (
        "Engagement Improvement",
        "Engagement_Improvement"
    ),

    (
        "Retention Percentage",
        "Retention_Percent_of_Post"
    )
]


between_rows = []

for label, variable in between_variables:

    ai = df[
        df["Group"] == "AI"
    ][variable].dropna()

    control = df[
        df["Group"] == "Control"
    ][variable].dropna()


    ai_normal = False
    control_normal = False

    if len(ai) >= 3:

        ai_shapiro_p = shapiro(ai).pvalue
        ai_normal = ai_shapiro_p > 0.05

    else:

        ai_shapiro_p = np.nan

    if len(control) >= 3:

        control_shapiro_p = shapiro(control).pvalue
        control_normal = control_shapiro_p > 0.05

    else:

        control_shapiro_p = np.nan


    if ai_normal and control_normal:

        statistic, p_value = ttest_ind(
            ai,
            control,
            equal_var=False
        )

        test_name = "Welch independent t-test"

        effect = cohens_d_independent(
            ai,
            control
        )

        effect_name = "Cohen's d"

    else:

        statistic, p_value = mannwhitneyu(
            ai,
            control,
            alternative="two-sided"
        )

        test_name = "Mann-Whitney U"

        effect = rank_biserial_from_u(
            statistic,
            len(ai),
            len(control)
        )

        effect_name = "Rank-biserial correlation"

    between_rows.append({

        "Outcome": label,
        "Variable": variable,

        "AI_N": len(ai),
        "AI_Mean": ai.mean(),
        "AI_SD": ai.std(ddof=1),

        "Control_N": len(control),
        "Control_Mean": control.mean(),
        "Control_SD": control.std(ddof=1),

        "AI_Shapiro_p": ai_shapiro_p,
        "Control_Shapiro_p": control_shapiro_p,

        "Test": test_name,
        "Statistic": statistic,
        "p_value": p_value,

        "Effect_Size": effect,
        "Effect_Size_Type": effect_name
    })


between_results = pd.DataFrame(
    between_rows
)



from statsmodels.stats.multitest import multipletests

if len(between_results) > 0:

    between_results[
        "Holm_adjusted_p"
    ] = multipletests(
        between_results["p_value"],
        method="holm"
    )[1]


print("\nBetween-group results:")
display(
    between_results.round(4)
)

between_results.to_csv(
    "between_group_results.csv",
    index=False
)



print("\n============================================================")
print("WITHIN-GROUP PAIRED ANALYSIS")
print("============================================================")


paired_comparisons = [

    (
        "Score",
        "Pre_Score_10",
        "Post_Score_10",
        "Pre → Post"
    ),

    (
        "Score",
        "Post_Score_10",
        "Retention_Score_10",
        "Post → Retention"
    ),

    (
        "Completion Time",
        "Pre_Time_sec",
        "Post_Time_sec",
        "Pre → Post"
    ),

    (
        "Completion Time",
        "Post_Time_sec",
        "Retention_Time_sec",
        "Post → Retention"
    ),

    (
        "Engagement",
        "Pre_Engagement_1to5",
        "Post_Engagement_1to5",
        "Pre → Post"
    ),

    (
        "Engagement",
        "Post_Engagement_1to5",
        "Retention_Engagement_1to5",
        "Post → Retention"
    )
]


within_rows = []


for group in ["AI", "Control"]:

    group_data = df[
        df["Group"] == group
    ]

    for outcome, before_col, after_col, comparison in paired_comparisons:

        temp = group_data[
            [
                before_col,
                after_col
            ]
        ].dropna()

        before = temp[
            before_col
        ]

        after = temp[
            after_col
        ]

        differences = after - before

        if len(differences) >= 3:

            shapiro_diff_p = shapiro(
                differences
            ).pvalue

        else:

            shapiro_diff_p = np.nan




        if (
            not np.isnan(shapiro_diff_p)
            and
            shapiro_diff_p > 0.05
        ):

            statistic, p_value = ttest_rel(
                before,
                after
            )

            test_name = "Paired t-test"

            effect = cohens_dz(
                before,
                after
            )

            effect_name = "Cohen's dz"

        else:

            statistic, p_value = wilcoxon(
                before,
                after
            )

            test_name = "Wilcoxon signed-rank"


            effect = cohens_dz(
                before,
                after
            )

            effect_name = "Cohen's dz"

        within_rows.append({

            "Group": group,
            "Outcome": outcome,
            "Comparison": comparison,

            "N": len(temp),

            "Before_Mean":
                before.mean(),

            "After_Mean":
                after.mean(),

            "Mean_Change":
                differences.mean(),

            "SD_Change":
                differences.std(ddof=1),

            "Shapiro_p_difference":
                shapiro_diff_p,

            "Test":
                test_name,

            "Statistic":
                statistic,

            "p_value":
                p_value,

            "Effect_Size":
                effect,

            "Effect_Size_Type":
                effect_name
        })


within_results = pd.DataFrame(
    within_rows
)



within_results[
    "Holm_adjusted_p"
] = multipletests(
    within_results["p_value"],
    method="holm"
)[1]


print("\nWithin-group results:")
display(
    within_results.round(4)
)

within_results.to_csv(
    "within_group_results.csv",
    index=False
)



print("\n============================================================")
print("ANCOVA")
print("============================================================")


df["AI_indicator"] = (
    df["Group"] == "AI"
).astype(int)


if df["AI_indicator"].nunique() < 2:

    print(
        "ANCOVA cannot be performed because "
        "only one group is present."
    )

    ancova_results = pd.DataFrame()

else:


    model = ols(
        "Post_Score_10 ~ AI_indicator + Pre_Score_10",
        data=df
    ).fit(
        cov_type="HC3"
    )

    ci = model.conf_int().loc[
        "AI_indicator"
    ]

    ancova_results = pd.DataFrame([{

        "Outcome":
            "Post-test score adjusted for pre-test",

        "Adjusted_AI_minus_Control":
            model.params["AI_indicator"],

        "Robust_SE":
            model.bse["AI_indicator"],

        "CI_95_lower":
            ci.iloc[0],

        "CI_95_upper":
            ci.iloc[1],

        "p_value":
            model.pvalues["AI_indicator"],

        "R_squared":
            model.rsquared

    }])

    display(
        ancova_results.round(4)
    )

    ancova_results.to_csv(
        "ancova_results.csv",
        index=False
    )

    print("\nFull ANCOVA model:")
    print(model.summary())



print("\n============================================================")
print("PRE-POST CORRELATIONS")
print("============================================================")


correlation_rows = []

for group in ["AI", "Control"]:

    subset = df[
        df["Group"] == group
    ]

    temp = subset[
        [
            "Pre_Score_10",
            "Post_Score_10"
        ]
    ].dropna()

    if len(temp) >= 3:

        pearson_r, pearson_p = stats.pearsonr(
            temp["Pre_Score_10"],
            temp["Post_Score_10"]
        )

        spearman_r, spearman_p = stats.spearmanr(
            temp["Pre_Score_10"],
            temp["Post_Score_10"]
        )

    else:

        pearson_r = np.nan
        pearson_p = np.nan
        spearman_r = np.nan
        spearman_p = np.nan


    correlation_rows.append({

        "Group": group,

        "Pearson_r":
            pearson_r,

        "Pearson_p":
            pearson_p,

        "Spearman_rho":
            spearman_r,

        "Spearman_p":
            spearman_p
    })


correlation_results = pd.DataFrame(
    correlation_rows
)

display(
    correlation_results.round(4)
)

correlation_results.to_csv(
    "correlations.csv",
    index=False
)



print("\nCreating graphs...")


score_long = df.melt(

    id_vars=[
        "Participant_Code",
        "Group"
    ],

    value_vars=[
        "Pre_Score_10",
        "Post_Score_10",
        "Retention_Score_10"
    ],

    var_name="Timepoint",
    value_name="Score"
)


score_long["Timepoint"] = (
    score_long["Timepoint"]
    .map({
        "Pre_Score_10": "Pre-test",
        "Post_Score_10": "Post-test",
        "Retention_Score_10": "Retention"
    })
)


timepoint_order = [
    "Pre-test",
    "Post-test",
    "Retention"
]


plt.figure(
    figsize=(10, 6)
)

sns.pointplot(
    data=score_long,
    x="Timepoint",
    y="Score",
    hue="Group",
    order=timepoint_order,
    errorbar="sd",
    dodge=False
)

plt.title(
    "Mean Test Scores by Group and Timepoint"
)

plt.xlabel("Assessment Timepoint")
plt.ylabel("Score (0–10)")

plt.ylim(0, 10)

plt.legend(
    title="Group"
)

plt.tight_layout()

plt.savefig(
    "01_score_trajectory.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



plt.figure(
    figsize=(8, 6)
)

sns.boxplot(
    data=df,
    x="Group",
    y="Score_Improvement"
)

sns.stripplot(
    data=df,
    x="Group",
    y="Score_Improvement",
    color="black",
    alpha=0.6
)

plt.axhline(
    0,
    linestyle="--"
)

plt.title(
    "Post-test Score Improvement"
)

plt.xlabel("Group")

plt.ylabel(
    "Score Improvement\n(Post-test − Pre-test)"
)

plt.tight_layout()

plt.savefig(
    "02_score_improvement.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



time_long = df.melt(

    id_vars=[
        "Participant_Code",
        "Group"
    ],

    value_vars=[
        "Pre_Time_sec",
        "Post_Time_sec",
        "Retention_Time_sec"
    ],

    var_name="Timepoint",
    value_name="Completion_Time_sec"
)


time_long["Timepoint"] = (
    time_long["Timepoint"]
    .map({
        "Pre_Time_sec": "Pre-test",
        "Post_Time_sec": "Post-test",
        "Retention_Time_sec": "Retention"
    })
)


plt.figure(
    figsize=(10, 6)
)

sns.pointplot(
    data=time_long,
    x="Timepoint",
    y="Completion_Time_sec",
    hue="Group",
    order=timepoint_order,
    errorbar="sd",
    dodge=False
)

plt.title(
    "Mean Completion Time by Group and Timepoint"
)

plt.xlabel("Assessment Timepoint")

plt.ylabel(
    "Completion Time (seconds)"
)

plt.legend(
    title="Group"
)

plt.tight_layout()

plt.savefig(
    "03_completion_time.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



plt.figure(
    figsize=(8, 6)
)

sns.boxplot(
    data=df,
    x="Group",
    y="Time_Reduction_sec"
)

sns.stripplot(
    data=df,
    x="Group",
    y="Time_Reduction_sec",
    color="black",
    alpha=0.6
)

plt.axhline(
    0,
    linestyle="--"
)

plt.title(
    "Reduction in Completion Time"
)

plt.xlabel("Group")

plt.ylabel(
    "Time Reduction (seconds)\n"
    "Pre-test − Post-test"
)

plt.tight_layout()

plt.savefig(
    "04_time_reduction.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



engagement_long = df.melt(

    id_vars=[
        "Participant_Code",
        "Group"
    ],

    value_vars=[
        "Pre_Engagement_1to5",
        "Post_Engagement_1to5",
        "Retention_Engagement_1to5"
    ],

    var_name="Timepoint",
    value_name="Engagement"
)


engagement_long["Timepoint"] = (
    engagement_long["Timepoint"]
    .map({
        "Pre_Engagement_1to5": "Pre-test",
        "Post_Engagement_1to5": "Post-test",
        "Retention_Engagement_1to5": "Retention"
    })
)


plt.figure(
    figsize=(10, 6)
)

sns.pointplot(
    data=engagement_long,
    x="Timepoint",
    y="Engagement",
    hue="Group",
    order=timepoint_order,
    errorbar="sd",
    dodge=False
)

plt.title(
    "Mean Engagement by Group and Timepoint"
)

plt.xlabel("Assessment Timepoint")

plt.ylabel(
    "Engagement Rating (1–5)"
)

plt.ylim(1, 5)

plt.legend(
    title="Group"
)

plt.tight_layout()

plt.savefig(
    "05_engagement.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



retention_long = df.melt(

    id_vars=[
        "Participant_Code",
        "Group"
    ],

    value_vars=[
        "Post_Score_10",
        "Retention_Score_10"
    ],

    var_name="Timepoint",
    value_name="Score"
)


retention_long["Timepoint"] = (
    retention_long["Timepoint"]
    .map({
        "Post_Score_10": "Post-test",
        "Retention_Score_10": "Retention"
    })
)


plt.figure(
    figsize=(9, 6)
)

sns.pointplot(
    data=retention_long,
    x="Timepoint",
    y="Score",
    hue="Group",
    order=[
        "Post-test",
        "Retention"
    ],
    errorbar="sd",
    dodge=False
)

plt.title(
    "Post-test to Retention Score"
)

plt.xlabel("Assessment")

plt.ylabel("Score (0–10)")

plt.ylim(0, 10)

plt.legend(
    title="Group"
)

plt.tight_layout()

plt.savefig(
    "06_retention.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


plt.figure(
    figsize=(12, 7)
)

for group in ["AI", "Control"]:

    group_data = df[
        df["Group"] == group
    ]

    for _, row in group_data.iterrows():

        plt.plot(
            [
                "Pre-test",
                "Post-test",
                "Retention"
            ],

            [
                row["Pre_Score_10"],
                row["Post_Score_10"],
                row["Retention_Score_10"]
            ],

            alpha=0.25
        )


for group in ["AI", "Control"]:

    group_data = df[
        df["Group"] == group
    ]

    means = [
        group_data["Pre_Score_10"].mean(),
        group_data["Post_Score_10"].mean(),
        group_data["Retention_Score_10"].mean()
    ]

    plt.plot(
        [
            "Pre-test",
            "Post-test",
            "Retention"
        ],
        means,
        linewidth=3,
        marker="o",
        label=f"{group} Mean"
    )


plt.title(
    "Individual Participant Score Trajectories"
)

plt.xlabel(
    "Assessment Timepoint"
)

plt.ylabel(
    "Score (0–10)"
)

plt.ylim(0, 10)

plt.legend()

plt.tight_layout()

plt.savefig(
    "07_individual_score_trajectories.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



plt.figure(
    figsize=(9, 6)
)

sns.histplot(
    data=df,
    x="Post_Score_10",
    hue="Group",
    kde=True,
    bins=10,
    element="step"
)

plt.title(
    "Distribution of Post-test Scores"
)

plt.xlabel(
    "Post-test Score (0–10)"
)

plt.ylabel(
    "Number of Participants"
)

plt.tight_layout()

plt.savefig(
    "08_post_score_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



plt.figure(
    figsize=(8, 7)
)

sns.scatterplot(
    data=df,
    x="Pre_Score_10",
    y="Post_Score_10",
    hue="Group",
    s=80
)

plt.plot(
    [0, 10],
    [0, 10],
    linestyle="--"
)

plt.title(
    "Pre-test vs Post-test Scores"
)

plt.xlabel(
    "Pre-test Score"
)

plt.ylabel(
    "Post-test Score"
)

plt.xlim(0, 10)

plt.ylim(0, 10)

plt.legend(
    title="Group"
)

plt.tight_layout()

plt.savefig(
    "09_pre_vs_post_score.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



plt.figure(
    figsize=(8, 6)
)

sns.boxplot(
    data=df,
    x="Group",
    y="Retention_Percent_of_Post"
)

sns.stripplot(
    data=df,
    x="Group",
    y="Retention_Percent_of_Post",
    color="black",
    alpha=0.6
)

plt.axhline(
    100,
    linestyle="--"
)

plt.title(
    "Retention as Percentage of Post-test Score"
)

plt.xlabel("Group")

plt.ylabel(
    "Retention (% of Post-test)"
)

plt.tight_layout()

plt.savefig(
    "10_retention_percentage.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



print("\n============================================================")
print("FINAL ANALYSIS SUMMARY")
print("============================================================")


final_summary = []


for group in ["AI", "Control"]:

    subset = df[
        df["Group"] == group
    ]

    final_summary.append({

        "Group": group,

        "N":
            len(subset),

        "Pre_Score_Mean":
            subset["Pre_Score_10"].mean(),

        "Post_Score_Mean":
            subset["Post_Score_10"].mean(),

        "Retention_Score_Mean":
            subset["Retention_Score_10"].mean(),

        "Score_Improvement_Mean":
            subset["Score_Improvement"].mean(),

        "Pre_Time_Mean":
            subset["Pre_Time_sec"].mean(),

        "Post_Time_Mean":
            subset["Post_Time_sec"].mean(),

        "Time_Reduction_Mean":
            subset["Time_Reduction_sec"].mean(),

        "Pre_Engagement_Mean":
            subset["Pre_Engagement_1to5"].mean(),

        "Post_Engagement_Mean":
            subset["Post_Engagement_1to5"].mean(),

        "Engagement_Improvement_Mean":
            subset["Engagement_Improvement"].mean(),

        "Retention_Percent_Mean":
            subset["Retention_Percent_of_Post"].mean()
    })


final_summary = pd.DataFrame(
    final_summary
)

display(
    final_summary.round(3)
)

final_summary.to_csv(
    "final_summary.csv",
    index=False
)



readme_text = """

THESIS DATA ANALYSIS OUTPUT
===========================

Dataset:
50 participant dataset

Groups:
AI
Control

Expected sample:
25 AI
25 Control

Main outcomes:
1. Pre-test score
2. Post-test score
3. Retention score
4. Completion time
5. Engagement

Statistical analyses:
- Descriptive statistics
- Shapiro-Wilk normality test
- Welch independent t-test where appropriate
- Mann-Whitney U test where appropriate
- Paired t-test where appropriate
- Wilcoxon signed-rank test where appropriate
- Cohen's d
- Cohen's dz
- Rank-biserial correlation
- Holm multiple-comparison correction
- ANCOVA adjusting post-test score for pre-test score
- Pearson and Spearman correlations

Graphs:
01_score_trajectory.png
02_score_improvement.png
03_completion_time.png
04_time_reduction.png
05_engagement.png
06_retention.png
07_individual_score_trajectories.png
08_post_score_distribution.png
09_pre_vs_post_score.png
10_retention_percentage.png


"""

with open(
    "README_analysis.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write(readme_text)


excel_output = "thesis_analysis_results.xlsx"

with pd.ExcelWriter(
    excel_output,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned_Data",
        index=False
    )

    descriptive_results.to_excel(
        writer,
        sheet_name="Descriptive",
        index=False
    )

    normality_results.to_excel(
        writer,
        sheet_name="Normality",
        index=False
    )

    between_results.to_excel(
        writer,
        sheet_name="Between_Group",
        index=False
    )

    within_results.to_excel(
        writer,
        sheet_name="Within_Group",
        index=False
    )

    if len(ancova_results) > 0:

        ancova_results.to_excel(
            writer,
            sheet_name="ANCOVA",
            index=False
        )

    correlation_results.to_excel(
        writer,
        sheet_name="Correlations",
        index=False
    )

    final_summary.to_excel(
        writer,
        sheet_name="Final_Summary",
        index=False
    )

print("\nExcel report created:")
print(excel_output)



output_folder = "thesis_analysis_outputs"

os.makedirs(
    output_folder,
    exist_ok=True
)


import shutil

files_to_copy = [

    "cleaned_thesis_dataset.csv",
    "descriptive_statistics.csv",
    "group_summary.csv",
    "normality_tests.csv",
    "between_group_results.csv",
    "within_group_results.csv",
    "correlations.csv",
    "final_summary.csv",
    "thesis_analysis_results.xlsx",
    "README_analysis.txt",

    "01_score_trajectory.png",
    "02_score_improvement.png",
    "03_completion_time.png",
    "04_time_reduction.png",
    "05_engagement.png",
    "06_retention.png",
    "07_individual_score_trajectories.png",
    "08_post_score_distribution.png",
    "09_pre_vs_post_score.png",
    "10_retention_percentage.png"
]


for filename in files_to_copy:

    if os.path.exists(filename):

        shutil.copy(
            filename,
            output_folder
        )



zip_filename = "THESIS_COMPLETE_ANALYSIS.zip"

with zipfile.ZipFile(
    zip_filename,
    "w",
    zipfile.ZIP_DEFLATED
) as zipf:

    for root, dirs, files_in_folder in os.walk(
        output_folder
    ):

        for filename in files_in_folder:

            filepath = os.path.join(
                root,
                filename
            )

            arcname = os.path.relpath(
                filepath,
                output_folder
            )

            zipf.write(
                filepath,
                arcname
            )



print("\n")
print("============================================================")
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("============================================================")

print("\nParticipant counts:")
print(df["Group"].value_counts())

print("\nGenerated files:")
print(
    f"- {excel_output}"
)

print(
    f"- {zip_filename}"
)

print(
    "\nThe ZIP contains the statistical tables, "
    "Excel report, CSV files, graphs, and README."
)

print("\nIMPORTANT:")
print(
    "This is the results in thesis "

)



files.download(
    zip_filename
)
```
